Utiliza la variable content para crear la sopa. En la sintaxis de la sopa usa la variable content cómo si fuera response.content

```python
# Convertir a Beautiful soup

sopa_rutas= BeautifulSoup(content, 'html.parser')

# Resultados del objeto de Beautiful Soup. 
#print(sopa_rutas.prettify())
```

```python
# Extracción de información 

# 1. Creación diccionario vacio donde apendeo los textos encontrados
diccionario_rutas = {'Nombre':[], 'Lugar':[], 'Tipo_ruta':[],'Duración':[],'Km':[], 'Dificultad':[],'Esfuerzo':[],'Descripción': [] }


#2. Extracción de información por categorias 
# Nombre rutas. Búsqueda de etiqueta y appendeo en lista, extracción de texto con getText()
lista_nombres= sopa_rutas.find_all("h3", {"class": "elementor-heading-title elementor-size-default"})

for nombre in lista_nombres[2:]:
    diccionario_rutas['Nombre'].append(nombre.getText(strip=True))  
 

 #4. Contenido en Datos técnicos

 # Extraemos informacion de la etiqueta li, ya que se refiere a detalles técnicos donde sacaremos lu
li_general= sopa_rutas.find_all('li')

# Iterar sobre cada li ya que contiene los elementos que necesitamos
for li in li_general:
    
    
    #Lugar donde se realiza
    if "Donde se realizan:" in li.get_text():
        # Extraer el texto completo
        texto_lugar = li.get_text(strip=True)
        
        # Quitamos la parte "Donde se realizan:" 
        lugar = texto_lugar.replace("Donde se realizan:", "").strip()

        diccionario_rutas['Lugar'].append(lugar)   

    # Tipo de ruta
    if "Tipo de ruta:" in li.get_text():
        # Extraer el texto de tipo de ruta
        texto_ruta = li.get_text(strip=True)
        
        # Quitamos la parte "Tipo de ruta:" 
        ruta= texto_ruta.replace("Tipo de ruta:", "").strip()
        diccionario_rutas['Tipo_ruta'].append(ruta)   
    

    # Duracion
    if "Duración:" in li.get_text():
        # Extraer el texto de duracion
        texto_duracion= li.get_text(strip=True)
        
        # Quitamos la parte "Duración:" 
        duracion= texto_duracion.replace("Duración:", "").strip()
        diccionario_rutas['Duración'].append(duracion)

     # Kilometros
     #Primero sacaré el texto ya que tenemos Km y km 
    texto_km = li.get_text(strip=True)

    # Extraemos texto
    if "Km" in texto_km or 'km' in texto_km: 
            # Quitamos la parte "km:" 
            km =  texto_km.replace("Km", "").replace("km", "").strip(':').strip()
            diccionario_rutas['Km'].append(km)


    # Dificultad
    #  #Primero sacaré el texto ya que tenemos Dificultad Técnica: y "Dificultad técnica:
    
    texto_dificultad = li.get_text(strip=True)
    # Dificultad
    if "Dificultad Técnica:" in texto_dificultad or "Dificultad técnica:" in texto_dificultad :
        dificultad = texto_dificultad.replace("Dificultad Técnica:", "").replace("Dificultad técnica:", "").strip()
        diccionario_rutas['Dificultad'].append(dificultad)

    # Esfuerzo
    if "Esfuerzo:" in li.get_text():
        # Extraer el texto de esfuerzo
        texto_esfuerzo = li.get_text(strip=True)
        
        # Quitamos la parte "Esfuerzo:" 
        esfuerzo= texto_esfuerzo.replace("Esfuerzo:", ""). replace(".", "").strip()
        diccionario_rutas['Esfuerzo'].append(esfuerzo)

#Descripcion
# Identificacion de ID para cada elento   
# Encontrar todos los paragrafois
 # Iterar sobre cada ID en la lista

Lista_ID1= ['1033', '1393', '2583', '1213', '9793','1583', '1843', '9713', '2533', '1453' ]

for id in Lista_ID1:
    # Encontrar todos los parágrafos correspondientes a este ID
    paragrafos = sopa_rutas.find_all("div", {"id": f"elementor-tab-content-{id}"})
    
    # Inicializar una variable para acumular el texto de todos los parágrafos de este ID
    texto_completo = ""

    for ruta_text in paragrafos:
        # Extraer el texto de cada parágrafo y agregarlo a la variable texto_completo
        texto_completo += ruta_text.get_text(strip=True) + " "  # Añadir un espacio entre parágrafos si es necesario
    
    # Agregar el texto completo al diccionario bajo la clave 'Descripción'
    if texto_completo:
        diccionario_rutas['Descripción'].append(texto_completo.strip())  # Elimina espacios innecesarios








    

```
